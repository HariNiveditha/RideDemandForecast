```python
import pandas as pd
import numpy as np
import gc
```

```python
base_path = "/content/drive/MyDrive/RideDemandForecast"

files = {
    "04": f"{base_path}/fhvhv_tripdata_2021-04.parquet",
    "05": f"{base_path}/fhvhv_tripdata_2021-05.parquet",
    "06": f"{base_path}/fhvhv_tripdata_2021-06.parquet",
}
```

```python
def process_month(file_path, month):

    print(f"\nProcessing {month}...")

    # Load parquet
    df = pd.read_parquet(file_path)

    print("Original Shape:", df.shape)

    # Keep only required columns
    df = df[
        [
            "pickup_datetime",
            "PULocationID"
        ]
    ]

    # Remove missing values
    df.dropna(inplace=True)

    # Convert datetime
    df["pickup_datetime"] = pd.to_datetime(df["pickup_datetime"])

    # Round to nearest hour
    df["pickup_hour"] = df["pickup_datetime"].dt.floor("h")

    # Aggregate ride demand
    hourly = (
        df.groupby(["pickup_hour", "PULocationID"])
        .size()
        .reset_index(name="ride_count")
    )

    print("Aggregated Shape:", hourly.shape)

    # Save
    output_file = f"{base_path}/hourly_demand_{month}.parquet"
    hourly.to_parquet(output_file, index=False)

    print("Saved:", output_file)

    # Free memory
    del df
    del hourly
    gc.collect()
```

```python
process_month(files["04"], "04")
```

```python
process_month(files["05"], "05")
```

```python
process_month(files["06"], "06")
```
